# Hàm tính IOU giữa 2 bbox

```python
def get_iou(boxA, boxB):
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    interArea = max(0, xB - xA) * max(0, yB - yA)
    boxAArea = (boxA[2]-boxA[0])*(boxA[3]-boxA[1])
    boxBArea = (boxB[2]-boxB[0])*(boxB[3]-boxB[1])
    iou = interArea / float(boxAArea + boxBArea - interArea + 1e-6)
    return iou
```

# Phân tích trạng thái ghế

```python
def check_chair_occupancy(detections, iou_threshold=0.2):
    """
    detections: list của YOLO output per frame
    Trả về dict {chair_id: {'status': 'VACANT'/'OCCUPIED', 'box': [x1,y1,x2,y2]}}
    """
    chairs = [d for d in detections if d['class_name']=='chair']
    persons = [d for d in detections if d['class_name']=='person']

    chair_statuses = {}
    for idx, chair in enumerate(chairs):
        chair_box = chair['box']
        is_occupied = False
        for person in persons:
            if get_iou(chair_box, person['box']) > iou_threshold:
                is_occupied = True
                break
        chair_statuses[idx] = {
            'status': 'OCCUPIED' if is_occupied else 'VACANT',
            'box': chair_box
        }
    return chair_statuses
```

```python
import cv2
from ultralytics import YOLO
import numpy as np

def get_iou(boxA, boxB):
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])
    interW = max(0, xB - xA)
    interH = max(0, yB - yA)
    interArea = interW * interH
    boxAArea = max(0, (boxA[2]-boxA[0]))*max(0,(boxA[3]-boxA[1]))
    boxBArea = max(0, (boxB[2]-boxB[0]))*max(0,(boxB[3]-boxB[1]))
    return interArea / float(boxAArea + boxBArea - interArea + 1e-6)

# ---- Load model ----
model = YOLO("best.pt")
cap = cv2.VideoCapture(0)
FPS = cap.get(cv2.CAP_PROP_FPS)
FRAME_DURATION = 1 / FPS if FPS > 0 else 1 / 30

# ---- Hyperparameters ----
IOU_THRESHOLD = 0.15
CHAIR_MARGIN_X = 15
CHAIR_MARGIN_Y_TOP = 30
CHAIR_MARGIN_Y_BOTTOM = 10
DEBOUNCE_FRAMES = 2  # giảm số frame debounce

time_tracker = []

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.track(frame, persist=True, tracker='bytetrack.yaml', conf=0.4, verbose=False)[0]

    detections = []
    if results.boxes is not None and len(results.boxes) > 0:
        boxes = results.boxes.xyxy.cpu().numpy()
        cls_ids = results.boxes.cls.cpu().numpy()
        for box, cls_id in zip(boxes, cls_ids):
            cls_name = model.names[int(cls_id)].lower()
            x1, y1, x2, y2 = map(int, box)
            detections.append({'box':[x1,y1,x2,y2],'class_name':cls_name})

    persons = [d for d in detections if d['class_name']=='person']
    chairs = [d for d in detections if d['class_name'] in ('chair','seat')]

    chair_status_list = []

    for chair in chairs:
        x1, y1, x2, y2 = chair['box']
        expanded_box = [
            x1 - CHAIR_MARGIN_X,
            y1 - CHAIR_MARGIN_Y_TOP,
            x2 + CHAIR_MARGIN_X,
            y2 + CHAIR_MARGIN_Y_BOTTOM
        ]

        is_occupied = False
        for p in persons:
            p_box = p['box']
            iou = get_iou(p_box, expanded_box)
            bottom_x = (p_box[0]+p_box[2])/2
            bottom_y = p_box[3]
            bottom_inside = (expanded_box[0]<=bottom_x<=expanded_box[2]) and (expanded_box[1]<=bottom_y<=expanded_box[3])
            if iou>=IOU_THRESHOLD or bottom_inside:
                is_occupied = True
                break

        chair_status_list.append({'box':expanded_box,'status':'OCCUPIED' if is_occupied else 'VACANT'})

    # ---- Cập nhật trạng thái & thời gian ----
    for ch in chair_status_list:
        matched = False
        for t in time_tracker:
            if get_iou(ch['box'], t['box']) > 0.5:
                # debounce trạng thái
                if t['status'] != ch['status']:
                    t['change_count'] += 1
                    if t['change_count'] >= DEBOUNCE_FRAMES:
                        t['status'] = ch['status']
                        t['change_count'] = 0
                else:
                    t['change_count'] = 0

                # đếm thời gian theo trạng thái mới ch['status']
                if ch['status'] == 'OCCUPIED':
                    t['OCCUPIED_FRAMES'] += 1
                else:
                    t['VACANT_FRAMES'] += 1

                # KHÔNG cập nhật bounding box, giữ box cũ
                matched = True
                break

        if not matched:
            # ghế mới, thêm vào tracker
            time_tracker.append({
                'box': ch['box'],
                'status': ch['status'],
                'OCCUPIED_FRAMES': 1 if ch['status']=='OCCUPIED' else 0,
                'VACANT_FRAMES': 1 if ch['status']=='VACANT' else 0,
                'change_count': 0
            })

    # ---- Vẽ ghế ----
    for t in time_tracker:
        x1, y1, x2, y2 = t['box']
        color = (0,0,255) if t['status']=='OCCUPIED' else (0,255,0)
        time_sec = t['OCCUPIED_FRAMES']*FRAME_DURATION if t['status']=='OCCUPIED' else t['VACANT_FRAMES']*FRAME_DURATION
        label = f"{t['status']} {time_sec:.1f}s"
        cv2.rectangle(frame,(x1,y1),(x2,y2),color,2)
        cv2.putText(frame,label,(x1,max(y1-10,20)),cv2.FONT_HERSHEY_SIMPLEX,0.6,color,2)

    # ---- Vẽ người ----
    for p in persons:
        x1, y1, x2, y2 = p['box']
        cv2.rectangle(frame,(x1,y1),(x2,y2),(255,0,0),2)
        cv2.putText(frame,"Person",(x1,max(y1-10,20)),cv2.FONT_HERSHEY_SIMPLEX,0.6,(255,0,0),2)

    cv2.imshow("Chair Detection with Time", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
